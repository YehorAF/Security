```python
from Crypto.Cipher import DES3

def pad(text):
    n = len(text) % 32
    return text + (b' ' * n)


key = b'123456781234567812345678'
text1 = b'123456781234567812345678'# + b" " * 30

des = DES3.new(key, DES3.MODE_ECB)

padded_text = pad(text1)
encrypted_text = des.encrypt(text1)

print(encrypted_text)
print(des.decrypt(encrypted_text))
```
